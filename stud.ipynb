```python
from sqlalchemy import create_engine, text
import pandas as pd

# 1. Create engine to connect to your database
engine = create_engine("mysql+mysqlconnector://root:23022006@localhost/STUDINFO") 

# Read the table into a DataFrame to verify
df = pd.read_sql("SELECT * FROM stud", con=engine)
print(df)
```

```python
# 2. Write the INSERT query and wrap it with text()
insert_query = text("""
    INSERT INTO stud (name, roll_no, cgpa)
    VALUES (:name, :roll_no, :cgpa)
""")

# 3. UPDATE operation
update_query = text("""
    UPDATE stud
    SET cgpa = :new_cgpa
    WHERE roll_no = :roll_no
""")

update_data = {
    "new_cgpa": 9.1,
    "roll_no": 105
}

with engine.connect() as connection:
    connection.execute(update_query, update_data)
    connection.commit() 
```
