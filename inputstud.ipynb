```python
from sqlalchemy import create_engine, text
import pandas as pd


# 1. Create engine to connect to your database
engine = create_engine("mysql+mysqlconnector://root:23022006@localhost/STUDINFO") 
```

```python
# 2. INSERT operation
name = input("Enter student name to insert: ")
roll_no = int(input("Enter roll number to insert: "))
cgpa = float(input("Enter CGPA to insert: "))

insert_query = text("""
    INSERT INTO stud (name, roll_no, cgpa)
    VALUES (:name, :roll_no, :cgpa)
""")

new_student = {
    "name": name,
    "roll_no": roll_no,
    "cgpa": cgpa
}

with engine.connect() as connection:
    connection.execute(insert_query, new_student)
    connection.commit()
    print("\nStudent inserted successfully.")

# 3. UPDATE operation
update_roll = int(input("\nEnter roll number of student to update CGPA: "))
new_cgpa = float(input("Enter new CGPA: "))

update_query = text("""
    UPDATE stud
    SET cgpa = :cgpa
    WHERE roll_no = :roll_no
""")

with engine.connect() as connection:
    connection.execute(update_query, {"cgpa": new_cgpa, "roll_no": update_roll})
    connection.commit()
    print("Student CGPA updated successfully.")

# 4. DELETE operation
delete_roll = int(input("\nEnter roll number of student to delete: "))

delete_query = text("""
    DELETE FROM stud
    WHERE roll_no = :roll_no
""")

with engine.connect() as connection:
    connection.execute(delete_query, {"roll_no": delete_roll})
    connection.commit()
    print("Student deleted successfully.")

```

```python
# 5. Display the final contents of the table
df = pd.read_sql("SELECT * FROM stud", con=engine)
print("\nFinal Student Table:\n", df)
```
