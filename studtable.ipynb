```python
from sqlalchemy import create_engine, text
import pandas as pd

# 1. Create engine to connect to your database
engine = create_engine("mysql+mysqlconnector://root:23022006@localhost/STUDINFO") 

# Read the table into a DataFrame to verify current records
df = pd.read_sql("SELECT * FROM stud", con=engine)
print("Before Insert and Delete:\n", df)
```

```python
# 2. INSERT query
insert_query = text("""
    INSERT INTO stud (name, roll_no, cgpa)
    VALUES (:name, :roll_no, :cgpa)
""")

new_student = {
    "name": "Billy",
    "roll_no": 112,
    "cgpa": 9.57
}

# Get roll number from user to delete
roll_to_delete = int(input("Enter the roll number to delete: "))

# 3. DELETE query to delete a specific student by roll number
delete_query = text("DELETE FROM stud WHERE roll_no = :roll_no")

with engine.connect() as connection:
    # Insert new record
    connection.execute(insert_query, new_student)
    
    # Delete record with specific roll number
    connection.execute(delete_query, {"roll_no": roll_to_delete})
    
    # Commit changes
    connection.commit()
```

```python
# Read and print table to verify
df = pd.read_sql("SELECT * FROM stud", con=engine)
print("After Delete:\n", df)
```
