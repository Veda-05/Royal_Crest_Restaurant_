```python
from sqlalchemy import create_engine, text
import pandas as pd

# 1. Create engine to connect to your database
engine = create_engine("mysql+mysqlconnector://root:23022006@localhost/STUDENTINFO") 

# Read the table into a DataFrame to verify
df = pd.read_sql("SELECT * FROM student", con=engine)
print(df)
```

```python
# 2. Write the INSERT query and wrap it with text()
insert_query = text("""
    INSERT INTO student (name, roll_no, cgpa)
    VALUES (:name, :roll_no, :cgpa)
""")

# 3. Data to insert
new_student = {
    "name": "Shree",
    "roll_no": 103,
    "cgpa": 9.27
}


with engine.connect() as connection:
    connection.execute(insert_query,new_student)
    connection.commit() 
```
